```python
import pandas as pd

# Carica i due file CSV
file1 = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\STAZIONI\STAZIONE6\6_rad_oraria.csv'  # Sostituisci con il percorso del tuo primo file CSV
file2 = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\STAZIONI\STAZIONE6\rad_extraterrestre_6.csv'  # Sostituisci con il percorso del tuo secondo file CSV

# Leggi i file CSV in DataFrame
df1 = pd.read_csv(file1,sep=';')
df2 = pd.read_csv(file2,sep=';')
print(df1)
print(df2)
# Rinomina la colonna "radiazione" in "radiazione1" per il primo DataFrame
df1 = df1.rename(columns={'radint': 'stazione'})

# Rinomina la colonna "radiazione" in "radiazione2" per il secondo DataFrame
df2 = df2.rename(columns={'radint': 'extraterrestre'})

# Unisci i due DataFrame in base alle colonne "data" e "ora"
merged_df = pd.merge(df1, df2, on=['data', 'ora'], how='inner')
print(merged_df)
# Salva il risultato in un nuovo file CSV
output_file = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\STAZIONI\STAZIONE6\6_extraterrestre_stazione.csv'  # Sostituisci con il nome desiderato per il file di output
merged_df.to_csv(output_file, index=False)

print(f"File CSV unito creato: {output_file}")
```
