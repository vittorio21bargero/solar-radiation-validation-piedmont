```python
import numpy as np
import matplotlib.pyplot as plt

import sqlite3

# Connessione al database SQLite
conn = sqlite3.connect(r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\radiazione.db')
cursor = conn.cursor()

# Query per selezionare i dati dalla prima colonna desiderata della tabella
query_colonna1 = "SELECT stazione129 FROM comparazione129"
cursor.execute(query_colonna1)
x = cursor.fetchall()

# Query per selezionare i dati dalla seconda colonna desiderata della tabella
query_colonna2 = "SELECT radint FROM comparazione129"
cursor.execute(query_colonna2)
y = cursor.fetchall()

# Chiusura della connessione al database
conn.close()

# Stampa dei dati ottenuti

# Converti le liste di tuple in array NumPy
# Converti le liste di tuple in array NumP
x = np.array([float(val[0]) for val in x ])
y = np.array([float(val[0]) for val in y ])


# Calcolo della retta di regressione lineare
slope, intercept = np.polyfit(x, y, 1)
line = slope * x + intercept

# Calcolo del coefficiente di correlazione lineare (r)
r = np.corrcoef(x, y)[0, 1]

# Creazione del grafico
plt.scatter(x, y, label='Dati')
plt.plot(x, line, color='red', label='Regressione lineare')

# Aggiunta del coefficiente di correlazione lineare (r) come testo nel grafico
plt.text(0.3, 4.3, f'Coefficiente di correlazione (r) = {r:.2f}', fontsize=10, color='blue')
plt.text(0.3, 30, f'pendenza retta = {slope:.2f}', fontsize=10, color='red')

# Aggiunta di titoli e legenda
plt.title('Correlazione lineare tra x e y')
plt.xlabel('stazione')
plt.ylabel('modello')
plt.legend()
plt.xlim(0,40)
plt.ylim(0,40)
# Mostra il grafico
plt.show()
```
