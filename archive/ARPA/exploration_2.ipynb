```python
import pandas as pd

def prepara_dati_giornalieri(file_path):
    # Leggi il file CSV
    df = pd.read_csv(file_path)

    # Converti la colonna 'data' in formato datetime
    df['data'] = pd.to_datetime(df['data'], format='%Y-%m-%d')

    
    
    # Sostituisci NaN con 0 per 'clearskyrad'
    df['clearskyrad'].fillna(0, inplace=True)

    # Funzione per sommare le colonne
    def somma_giornaliera(gruppo):
        if gruppo['nome'].iloc[0] == 'CASALE_MONFERRATO':
            return pd.Series({
                'rad': gruppo['rad'].sum(),
                'shortwave_radiation': gruppo['shortwave_radiation'].sum(),
                'clearskyrad': 0  # Assegna 0 per clearskyrad
            })
        else:
            return pd.Series({
                'rad': gruppo['rad'].sum(),
                'shortwave_radiation': gruppo['shortwave_radiation'].sum(),
                'clearskyrad': gruppo['clearskyrad'].sum()
            })

    # Raggruppa i dati per giorno e per nome, somma le colonne 'rad', 'shortwave_radiation', 'clearskyrad'
    df_giornaliero = df.groupby(['data', 'nome']).apply(somma_giornaliera).reset_index()

    # Ordina il DataFrame per nome e data
    df_giornaliero = df_giornaliero.sort_values(by=['nome', 'data'])
    df_giornaliero.to_csv(r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\ERA_PV_stazioni_clearsky_daily.csv', index=False, sep=',')
    return df_giornaliero

# Percorso del file CSV
file_path = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\ERA_PV_stazioni_clearsky.csv'  # Modifica con il tuo percorso

# Esegui la funzione
df_giornaliero = prepara_dati_giornalieri(file_path)

pd.set_option('display.max_rows', None) 
# Visualizza il risultato
print(df_giornaliero)
```
