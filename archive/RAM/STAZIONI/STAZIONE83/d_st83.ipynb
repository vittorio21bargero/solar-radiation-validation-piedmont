```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.dates as mdates

# Carica i dati dal file CSV
df = pd.read_csv(r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\DETERIORAMENTO_DATO\83_rad_daily_clearsky.csv', sep='|')

# Visualizza i primi 5 record per verificare che i dati siano corretti
print(df.head())

# Assicurati che la colonna 'data' sia convertita in un oggetto datetime
df['data'] = pd.to_datetime(df['data'], format='%Y-%m-%d')

# Plotta i dati della colonna 'stazione'
plt.figure(figsize=(10, 6))
plt.plot(df['data'], df['stazione'], marker='o')
plt.xlabel('Data')
plt.ylabel('Radiazione')
plt.grid(True)

# Imposta il formato dei tick dell'asse x per mostrare solo l'anno
ax = plt.gca()
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax.xaxis.set_major_locator(mdates.YearLocator())

plt.show()
```
