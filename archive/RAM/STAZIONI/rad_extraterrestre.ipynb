```python
import pvlib
import pandas as pd
import numpy as np

# Definisci la posizione (latitudine, longitudine e altitudine)
latitude = 44.75
longitude = 8.39
altitude = 250  # altitudine in metri

# Crea l'oggetto Location
location = pvlib.location.Location(latitude, longitude, altitude=altitude)

# Definisci l'intervallo di tempo in UTC
start_date = '2012-01-01 00:00:00'
end_date = '2023-12-31 23:00:00'
times = pd.date_range(start=start_date, end=end_date, freq='H', tz='UTC')

# Calcola la posizione solare per ogni ora
solar_position = location.get_solarposition(times)

# Calcola la radiazione extraterrestre
extra_radiation = pvlib.irradiance.get_extra_radiation(times)

# Calcola l'angolo zenitale del sole in radianti
zenith_angle_radians = np.radians(solar_position['apparent_zenith'])

# Calcola la radiazione extraterrestre orizzontale (GHI extra)
ghi_extra = extra_radiation * np.cos(zenith_angle_radians)

# Converti in orario locale
ghi_extra.index = ghi_extra.index.tz_convert('Europe/Rome')

# Imposta a zero i valori negativi (quando il sole è sotto l'orizzonte)
ghi_extra[ghi_extra < 0] = 0

# Crea un DataFrame con i risultati
data = pd.DataFrame({
    'DateTime': ghi_extra.index,  # Use the converted time
    'ExtraterrestrialRadiation': ghi_extra.values  # Use the calculated GHI
})

# Salva i risultati in un file CSV
output_file = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\STAZIONI\STAZIONE6\rad_extraterrestre_6.csv'
data.to_csv(output_file, index=False)

print(f"File CSV creato: {output_file}")

```
