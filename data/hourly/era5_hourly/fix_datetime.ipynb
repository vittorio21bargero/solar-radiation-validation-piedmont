```python
import os
import pandas as pd
from glob import glob

# Specifica il percorso della cartella contenente i file CSV
folder_path = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\ERA5 ORARI'  # Modifica con il percorso corretto

# Trova tutti i file CSV nella cartella
file_paths = glob(os.path.join(folder_path, '*.csv'))

# Itera su ogni file nella cartella
for file_path in file_paths:
    # Legge il file CSV
    df = pd.read_csv(file_path, delimiter=';', decimal='.')

    # Controlla se esiste almeno una colonna
    if df.shape[1] > 0:
        # Utilizza la prima colonna del DataFrame
        colonna_a = df.iloc[:, 0]

        # Converte i valori numerici in stringhe, riempiendo con zeri iniziali se necessario
        colonna_a_str = colonna_a.astype(str).str.zfill(10)

        # Estrai l'anno, il mese, il giorno e l'ora dalla stringa di 10 numeri
        df['anno'] = colonna_a_str.str[:4]
        df['mese'] = colonna_a_str.str[4:6]
        df['giorno'] = colonna_a_str.str[6:8]
        df['ORA'] = colonna_a_str.str[8:10]

        # Crea la nuova colonna 'data' nel formato 'dd/mm/yyyy'
        df['data'] = df['giorno'] + '/' + df['mese'] + '/' + df['anno']

        # Crea la nuova colonna 'ora' nel formato 'hh:00:00'
        df['ora'] = df['ORA'] + ':00:00'

        # Elimina la vecchia colonna con le stringhe da 10 numeri
        df.drop(df.columns[0], axis=1, inplace=True)  # Elimina la prima colonna

        # Elimina le colonne temporanee usate per la formattazione
        df.drop(['anno', 'mese', 'giorno', 'ORA'], axis=1, inplace=True)

        # Salva il file CSV modificato, sovrascrivendo l'originale
        df.to_csv(file_path, sep=';', decimal=',', index=False)

        print(f"File modificato: {file_path}")
    else:
        print(f"File vuoto o senza colonne: {file_path}")

print("Processo completato.")

```
