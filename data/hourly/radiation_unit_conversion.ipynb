```python
import os
import pandas as pd

# Percorso della cartella contenente i file CSV
cartella_csv = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\DATI ORARI'

# Itera su tutti i file nella cartella
for file_name in os.listdir(cartella_csv):
    if file_name.endswith('.csv'):
        # Percorso completo del file CSV
        file_path = os.path.join(cartella_csv, file_name)
        
        # Leggi il file CSV
        df = pd.read_csv(file_path, sep=';', decimal=',')  # specifica il separatore e il carattere decimale

        # Controlla se la colonna 'rad extrat' esiste nel DataFrame
        if 'rad extrat' in df.columns:
            # Moltiplica i valori della colonna 'rad extrat' per 3,6
            df['rad extrat'] = df['rad extrat'].apply(lambda x: x / 3.6)
            
            # Salva il file CSV modificato (sovrascrive l'originale)
            df.to_csv(file_path, sep=';', decimal=',', index=False)
            
            print(f"Modificato il file: {file_name}")
        else:
            print(f"La colonna 'rad extrat' non è presente nel file: {file_name}")
```
