```python
import pandas as pd
from datetime import datetime
import pytz

# Definizione del fuso orario di Roma e UTC
rome_tz = pytz.timezone('Europe/Rome')
utc_tz = pytz.UTC

# Funzione per convertire una data e ora da Roma a UTC
def convert_to_utc(row):
    try:
        # Combina data e ora e converte nel formato datetime
        local_time = rome_tz.localize(datetime.strptime(f"{row['data']} {row['ora']}", "%Y-%m-%d %H"))
        # Converte da Roma a UTC
        utc_time = local_time.astimezone(utc_tz)
        # Ritorna l'orario in formato HH:00
        return utc_time.strftime("%H")
    except Exception as e:
        print(f"Errore di conversione in riga {row.name}: {e}")
        return row['ora']  # Ritorna l'ora originale se si verifica un errore

# Funzione per processare un file CSV
def convert_single_file_to_utc(file_path):
    try:
        # Leggi il file CSV
        df = pd.read_csv(file_path, sep=',', decimal='.')

        # Converti l'orario da Roma a UTC
        for index, row in df.iterrows():
            df.at[index, 'ora'] = convert_to_utc(row)

            # Stampa il progresso ogni 10000 righe
            if index % 10000 == 0:
                print(f"Elaborata riga {index}")

        # Salva il file sovrascrivendo l'originale
        df.to_csv(file_path, index=False, sep=',', decimal='.')

        print(f"File '{file_path}' processato e orario convertito.")

    except Exception as e:
        print(f"Errore nel file '{file_path}': {e}")

# Esempio di utilizzo
file_path = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\PV.csv'  # Inserisci il percorso del file CSV
convert_single_file_to_utc(file_path)
```
