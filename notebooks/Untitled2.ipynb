```python
import pandas as pd

# Funzione per aggregare i dati giornalieri e creare un nuovo file CSV
def aggrega_dati_giornalieri(df, output_file):
    # Aggiungi una colonna 'data' come oggetto datetime senza l'ora
    df['data'] = pd.to_datetime(df['data'], dayfirst=True)

    # Raggruppa i dati per 'nome' e 'data', e somma 'clearskyrad' e 'rad' per ogni giorno
    df_aggregato = df.groupby(['nome', 'data']).agg({
        'clearskyrad': 'sum',
        'rad': 'sum',
        'shortwave_radiation': 'sum'
    }).reset_index()

    # Esporta il DataFrame aggregato in un nuovo file CSV senza la colonna 'ora'
    df_aggregato.to_csv(output_file, index=False)

    print(f"I dati aggregati giornalieri sono stati salvati in {output_file}")

# Carica il CSV originale
df = pd.read_csv('ERA_PV_stazioni_clearsky_9_15.csv')  # Inserisci il percorso corretto del tuo file CSV

# Definisci il nome del file di output
output_file = 'ERA_PV_stazioni_clearsky_giornalieri.csv'

# Esegui l'aggregazione e salva i dati in un nuovo file CSV
aggrega_dati_giornalieri(df, output_file)
```
