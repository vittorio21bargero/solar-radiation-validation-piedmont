```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Funzione per eseguire il fit lineare con dati aggregati per giorno
def fit_per_nome_giornaliero(df):
    risultati_fit = {}

    # Aggiungi una colonna 'data' in formato datetime per facilitare l'aggregazione
    df['data'] = pd.to_datetime(df['data'], dayfirst=True)

    # Raggruppa i dati per 'nome' e 'data' e somma 'clearskyrad' e 'rad' per ogni giorno
    df_aggregato = df.groupby(['nome', 'data']).agg({
        'clearskyrad': 'sum',
        'rad': 'sum'
    }).reset_index()

    # Raggruppa nuovamente i dati per 'nome'
    gruppi = df_aggregato.groupby('nome')

    for nome, gruppo in gruppi:
        # Escludi il nome 'CASALE_MONFERRATO'
        if nome == 'CASALE_MONFERRATO':
            continue
        
        # Estrai le colonne rad e clearskyrad per il fit
        X = gruppo['clearskyrad'].values.reshape(-1, 1)  # Variabile indipendente
        y = gruppo['rad'].values  # Variabile dipendente

        # Esegui il fit lineare
        modello = LinearRegression()
        modello.fit(X, y)

        # Salva i coefficienti del fit per ogni gruppo
        risultati_fit[nome] = {
            'coefficiente_angolare': modello.coef_[0],
            'intercetta': modello.intercept_,
            'R^2': modello.score(X, y)
        }

    return risultati_fit

# Carica il CSV
# Supponiamo che il CSV abbia colonne 'rad', 'clearsky_rad', 'data', 'nome'
df = pd.read_csv('ERA_PV_stazioni_clearsky.csv')  # Inserisci il percorso corretto del tuo file CSV

# Esegui il fit lineare per ogni gruppo con dati aggregati giornalieri
risultati = fit_per_nome_giornaliero(df)

# Visualizza i risultati
for nome, risultato in risultati.items():
    print(f"Fit per {nome}:")
    print(f"  Coefficiente angolare: {risultato['coefficiente_angolare']}")
    print(f"  Intercetta: {risultato['intercetta']}")
    print(f"  R^2: {risultato['R^2']}\n")
```
