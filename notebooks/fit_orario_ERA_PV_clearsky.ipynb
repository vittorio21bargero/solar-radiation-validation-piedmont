```python
import pandas as pd
import statsmodels.api as sm
import numpy as np

# Funzione per eseguire il fit lineare per ciascun gruppo
def fit_per_nome(df):
    risultati_fit = {}

    # Raggruppa i dati per 'nome'
    gruppi = df.groupby('nome')

    for nome, gruppo in gruppi:
        # Escludi il nome 'CASALE_MONFERRATO'
        if nome == 'CASALE_MONFERRATO':
            continue
        # Estrai le colonne rad e clearskyrad per il fit
        X = gruppo['clearskyrad'].values  # Variabile indipendente
        y = gruppo['shortwave_radiation'].values  # Variabile dipendente

        # Aggiungi una colonna di 1 per l'intercetta
        X_const = sm.add_constant(X)

        # Esegui il fit lineare con statsmodels
        modello = sm.OLS(y, X_const).fit()

        # Ottieni i risultati del fit
        coefficiente_angolare = modello.params[1]
        intercetta = modello.params[0]
        R_squared = modello.rsquared
        p_value_fisher = modello.f_pvalue

        # Calcola i valori previsti (y_pred) dal modello
        y_pred = modello.predict(X_const)

        # Calcola lo scarto percentuale medio tra y (osservato) e y_pred (previsto)
        scarto_percentuale = np.abs((y - y_pred) / y) * 100
        scarto_percentuale_medio = np.mean(scarto_percentuale)

        # Salva i coefficienti del fit e il p-value per ogni gruppo
        risultati_fit[nome] = {
            'coefficiente_angolare': coefficiente_angolare,
            'intercetta': intercetta,
            'R^2': R_squared,
            'p_value_fisher': p_value_fisher,
            'scarto_percentuale_medio': scarto_percentuale_medio
        }

    # Calcola lo scarto percentuale medio totale per l'intero DataFrame
    scarto_percentuale_totale = np.abs((df['clearskyrad'] - df['shortwave_radiation']) / df['clearskyrad']) * 100
    scarto_percentuale_medio_totale = np.mean(scarto_percentuale_totale)

    return risultati_fit, scarto_percentuale_medio_totale

# Carica il CSV
df = pd.read_csv('ERA_PV_stazioni_clearsky_9_15.csv')  # Inserisci il percorso corretto del tuo file CSV

# Esegui il fit lineare per ogni gruppo e calcola lo scarto percentuale medio totale
risultati, scarto_percentuale_medio_totale = fit_per_nome(df)

# Visualizza i risultati
for nome, risultato in risultati.items():
    print(f"Fit per {nome}:")
    print(f"  Coefficiente angolare: {risultato['coefficiente_angolare']}")
    print(f"  Intercetta: {risultato['intercetta']}")
    print(f"  R^2: {risultato['R^2']}")
    print(f"  p-value (Fisher): {risultato['p_value_fisher']}")
    print(f"  Scarto percentuale medio (y vs y_pred): {risultato['scarto_percentuale_medio']:.2f}%\n")

# Visualizza lo scarto percentuale medio totale
print(f"Scarto percentuale medio totale (x vs y): {scarto_percentuale_medio_totale:.2f}%")
```
