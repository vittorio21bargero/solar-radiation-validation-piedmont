```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Funzione per eseguire il fit lineare per ciascun gruppo
def fit_per_nome(df):
    risultati_fit = {}

    # Raggruppa i dati per 'nome'
    gruppi = df.groupby('nome')

    for nome, gruppo in gruppi:
        # Escludi il nome 'casale_monferrato'
        if nome == 'CASALE_MONFERRATO':
            continue
        # Estrai le colonne rad e clearsky_rad per il fit
        X = gruppo['clearskyrad'].values.reshape(-1, 1)  # Variabile indipendente
        y = gruppo['rad'].values  # Variabile dipendente

        # Esegui il fit lineare
        modello = LinearRegression()
        modello.fit(X, y)

        # Salva i coefficienti del fit per ogni gruppo
        risultati_fit[nome] = {
            'coefficiente_angolare': modello.coef_[0],
            'intercetta': modello.intercept_,
            'R^2': modello.score(X, y)
        }

    return risultati_fit

# Carica il CSV
# Supponiamo che il CSV abbia colonne 'rad', 'clearsky_rad', 'nome'
# Ad esempio: df = pd.read_csv('percorso_del_file.csv')
df = pd.read_csv('ERA_PV_stazioni_clearsky_9_15.csv')  # Inserisci il percorso corretto del tuo file CSV

# Esegui il fit lineare per ogni gruppo
risultati = fit_per_nome(df)

# Visualizza i risultati
for nome, risultato in risultati.items():
    print(f"Fit per {nome}:")
    print(f"  Coefficiente angolare: {risultato['coefficiente_angolare']}")
    print(f"  Intercetta: {risultato['intercetta']}")
    print(f"  R^2: {risultato['R^2']}\n")
```
