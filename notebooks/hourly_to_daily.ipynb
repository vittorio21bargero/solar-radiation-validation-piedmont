```python
import os
import pandas as pd

# Specifica il percorso della cartella di input e di output
input_folder = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\DATI ORARI'  # Sostituisci con il percorso della cartella contenente i file CSV
output_folder = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\DATI GIORNALIERI'  # Sostituisci con il percorso della cartella di output

# Verifica che la cartella di output esista, altrimenti la crea
os.makedirs(output_folder, exist_ok=True)

# Itera su tutti i file nella cartella di input
for file_name in os.listdir(input_folder):
    if file_name.endswith('.csv'):
        input_file = os.path.join(input_folder, file_name)
        output_file = os.path.join(output_folder, file_name)

        # Leggi il file CSV
        df = pd.read_csv(input_file, sep=';', decimal=',')

        # Converte la colonna 'data' in formato datetime
        df['data'] = pd.to_datetime(df['data'], format='mixed', dayfirst=True)

        # Converti le colonne 'rad' e 'rad extrat' in numerico, forzando gli errori a NaN e poi sostituendoli con 0
        df['rad'] = pd.to_numeric(df['rad'], errors='coerce').fillna(0)
        df['rad extrat'] = pd.to_numeric(df['rad extrat'], errors='coerce').fillna(0)

        # Raggruppa i dati per 'data' e somma i valori di 'rad' e 'rad extrat'
        df_summed = df.groupby('data')[['rad', 'rad extrat']].sum().reset_index()

        # Salva il risultato in un nuovo file CSV
        df_summed.to_csv(output_file, sep=';', decimal=',', index=False)

        print(f"File CSV creato: {output_file}")
```
