```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Funzione per eseguire il fit lineare per ciascun gruppo (nome + anno)
def fit_per_nome_e_anno(df):
    risultati_fit = []

    # Aggiungi una colonna 'anno' estraendola dalla colonna 'data'
    df['anno'] = pd.to_datetime(df['data']).dt.year

    # Raggruppa i dati per 'nome' e 'anno'
    gruppi = df.groupby(['nome', 'anno'])

    for (nome, anno), gruppo in gruppi:
        # Escludi il nome 'CASALE_MONFERRATO'
        if nome == 'CASALE_MONFERRATO':
            continue
        
        # Estrai le colonne rad e clearskyrad per il fit
        X = gruppo['clearskyrad'].values.reshape(-1, 1)  # Variabile indipendente
        y = gruppo['rad'].values  # Variabile dipendente

        # Esegui il fit lineare
        modello = LinearRegression()
        modello.fit(X, y)

        # Aggiungi il risultato alla lista con anno, nome, e R^2
        risultati_fit.append({
            'anno': anno,
            'stazione': nome,
            'R^2': modello.score(X, y)
        })

    # Converti i risultati in un DataFrame
    df_risultati = pd.DataFrame(risultati_fit)

    return df_risultati

# Carica il CSV
df = pd.read_csv('ERA_PV_stazioni_clearsky_9_15.csv')  # Inserisci il percorso corretto del tuo file CSV

# Esegui il fit lineare per ogni nome e per ogni anno
df_risultati = fit_per_nome_e_anno(df)
pd.set_option('display.max_rows', None)

# Stampa la tabella con anno, stazione e R^2
print(df_risultati)
```
