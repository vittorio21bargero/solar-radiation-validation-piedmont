```python
import pandas as pd

# Funzione per calcolare effective_cloudiness
def calcola_effective_cloudiness(row):
    return 0.7 * row['cloud_cover_low'] + 0.25 * row['cloud_cover_mid'] + 0.05 * row['cloud_cover_high']

# Carica il DataFrame
df = pd.read_csv('ERA_PV_stazioni.csv', delimiter=',')  # Modifica il nome del file e il delimitatore se necessario

# Calcola la colonna effective_cloudiness
df['effective_cloudiness'] = df.apply(calcola_effective_cloudiness, axis=1)

# Filtra i giorni con effective_cloudiness < 10 per ogni ora
giorni_validi = df.groupby('data').filter(lambda x: (x['effective_cloudiness'] < 10).all())

# Salva il DataFrame risultante in un nuovo file CSV
giorni_validi.to_csv('ERA_PV_stazioni_clearsky.csv', index=False)
```
