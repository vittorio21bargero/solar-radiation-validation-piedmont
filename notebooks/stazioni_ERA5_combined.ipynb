```python
import pandas as pd
import os

def process_files(first_folder, second_folder):
    # DataFrame vuoto per memorizzare i risultati
    combined_df = pd.DataFrame()

    # Itera su tutti i file della prima cartella
    for file1 in os.listdir(first_folder):
        if file1.endswith('.csv'):
            # Estrai il numero e la località dal nome del file della prima cartella
            localita = '_'.join(file1.split('_')[:-1])  # Ottieni il nome della località (es. roma)
            numero_localita = file1.split('_')[-1].replace('.csv', '')  # Estrai il numero della località (es. 12)
            file2 = f'era_{numero_localita}.csv'  # Nome del file corrispondente nella seconda cartella

            # Verifica che esista il file corrispondente nella seconda cartella
            if file2 in os.listdir(second_folder):
                # Costruisci i percorsi completi per i file da entrambe le cartelle
                file1_path = os.path.join(first_folder, file1)
                file2_path = os.path.join(second_folder, file2)

                # Carica i file CSV
                df1 = pd.read_csv(file1_path, sep=',', decimal='.', usecols=['data', 'ora', 'rad'])
                df2 = pd.read_csv(file2_path, sep=';', decimal='.', usecols=['data', 'ora', 'shortwave_radiation', 'cloud_cover_low', 'cloud_cover_mid', 'cloud_cover_high'])

                # Converti il formato della data in entrambi i DataFrame per unire correttamente
                df1['data'] = pd.to_datetime(df1['data'], format='%Y-%m-%d').dt.strftime('%Y-%m-%d')
                df2['data'] = pd.to_datetime(df2['data'], format='%d/%m/%Y').dt.strftime('%Y-%m-%d')

                # Unisci i due DataFrame sulle colonne 'data' e 'ora'
                merged_df = pd.merge(df1, df2, on=['data', 'ora'], how='inner')

                # Aggiungi la colonna 'localita' basata sul nome del file (es. roma)
                merged_df['nome'] = localita
                merged_df['localita'] = numero_localita

                # Concatena i dati uniti al DataFrame finale
                combined_df = pd.concat([combined_df, merged_df], ignore_index=True)

                print(f"Processato: {file1} e {file2}")

    # Ordina il DataFrame combinato prima per 'numero_localita', poi per 'localita', 'data' e 'ora'
    combined_df['data'] = pd.to_datetime(combined_df['data'])
    combined_df = combined_df.sort_values(by=['localita', 'nome', 'data', 'ora']).reset_index(drop=True)

    return combined_df

# Specifica i percorsi delle due cartelle
first_folder = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\DATI ORARI'
second_folder = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\ERA5 ORARI'

# Crea il DataFrame combinato a partire dai file delle due cartelle
final_df = process_files(first_folder, second_folder)

# Visualizza il DataFrame risultante
print("DataFrame finale:")
print(final_df.head())

# Salva il DataFrame combinato su un file CSV, se necessario
final_df.to_csv('ERA_stazioni.csv', index=False)
```
