```python
import pandas as pd

# Carica i due file CSV
file1 = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\PV.csv'  # File con colonne località, data, ora, clearskyrad
file2 = r'C:\Users\vitto\Desktop\VB\UNIVERSITÀ\STAGE\RADIAZIONE\ARPA\ERA_stazioni.csv'

print("Caricamento dei file...")

# Leggi i file CSV in DataFrame senza fare alcun ordinamento
df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

print("File caricati con successo. Formattazione delle colonne 'data' e 'ora'...")

# Assicurati che i formati delle colonne 'data' e 'ora' siano coerenti tra i due DataFrame
df1['data'] = pd.to_datetime(df1['data'], format='%Y-%m-%d')
df2['data'] = pd.to_datetime(df2['data'], format='%Y-%m-%d')

df1['ora'] = pd.to_datetime(df1['ora'], format='%H').dt.time
df2['ora'] = pd.to_datetime(df2['ora'], format='%H').dt.time

# Aggiungi una colonna indice al secondo DataFrame per mantenere l'ordine
df2['idx'] = df2.index

print("Formattazione completata. Avvio del merge batch...")

# Parametri per il batch
batch_size = 50000  # numero di righe per ogni batch
num_batches = len(df2) // batch_size + 1

merged_batches = []

# Processa df2 in batch per monitorare lo stato
for i in range(num_batches):
    start = i * batch_size
    end = (i + 1) * batch_size
    batch_df2 = df2.iloc[start:end]
    
    print(f"Unione batch {i+1}/{num_batches}... ({start} - {end} righe)")

    # Esegui il merge solo per il batch corrente
    merged_batch = pd.merge(batch_df2, df1, on=['localita', 'data', 'ora'], how='left')
    merged_batches.append(merged_batch)

# Unisci tutti i batch
merged_df = pd.concat(merged_batches)

print(f"Merge completato! Numero totale di righe: {len(merged_df)}. Riordinamento del DataFrame...")

# Riordina il DataFrame in base alla colonna 'idx' per preservare l'ordine del file 2
merged_df = merged_df.sort_values('idx').drop(columns=['idx'])

print("Riordinamento completato. Salvataggio del file...")

# Salva il risultato in un nuovo file CSV
output_file = 'ERA_PV_stazioni.csv'
merged_df.to_csv(output_file, index=False)

print(f"File unito creato con successo! Salvato come {output_file}.")
```
